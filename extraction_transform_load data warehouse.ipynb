```python
from sqlalchemy import create_engine
import pandas as pd

# 1. Konfigurasi detail koneksi PostgreSQL 
username = "postgres"  # default user postgres
password = "qwerty123"  # ganti dengan password database 
host = "localhost"  # ganti jika pakai cloud (misal URL Supabase/Neon)
port = "5432"  # port default postgresql
database = "db_ecommerce"  # nama database yang sudah dibuat

# 2. Membuat engine koneksi menggunakan format psycopg2
engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

print("Berhasil terhubung ke database PostgreSQL!")

# 3. Membaca 4 file CSV yang sudah bersih di Colab
df_customers = pd.read_csv("dim_customers.csv")
df_products = pd.read_csv("dim_products.csv")
df_orders = pd.read_csv("order.csv")
df_shipping = pd.read_csv("shipping_details.csv")

# 4. Memasukkan (ingest) data DataFrame ke dalam tabel-tabel PostgreSQL
df_customers.to_sql("dim_customers", engine, if_exists="replace", index=False)
df_products.to_sql("dim_products", engine, if_exists="replace", index=False)
df_orders.to_sql("order", engine, if_exists="replace", index=False)
df_shipping.to_sql("shipping", engine, if_exists="replace", index=False)

print("=" * 50)
print("SUKSES! Keempat tabel berhasil dimuat ke PostgreSQL.")
print("=" * 50)
```

```python
##input ulang data order dan shipping karena format tanggal bermasalah

import pandas as pd
from sqlalchemy import create_engine

# 1. Hubungkan ke database PostgreSQL (sesuaikan password)
engine = create_engine('postgresql://postgres:qwerty123@localhost:5432/db_ecommerce')

# 2. Baca file CSV
df_orders = pd.read_csv('order.csv')
df_shipping = pd.read_csv('shipping_details.csv')

# 3. Ubah fromat tanggal
df_orders['order_dt'] = pd.to_datetime(df_orders['order_date'], errors='coerce')
df_orders['order_dt'] = df_orders['order_dt'].fillna(pd.to_datetime(df_orders['order_date'], format='%d/%m/%Y', errors='coerce'))

df_orders_clean = df_orders[df_orders['order_dt'] <= '2026-09-30'].copy()
df_orders_clean = df_orders_clean.drop(columns=['order_dt'])

# Sesuaikan juga shipping_details agar hanya mengambil order_id yang valid
valid_order_ids = df_orders_clean['order_id']
df_shipping_clean = df_shipping[df_shipping['order_id'].isin(valid_order_ids)].copy()

# 4. Masukkan ke database dengan nama tabel baru
df_orders_clean.to_sql('orders_final', engine, if_exists='replace', index=False)
df_shipping_clean.to_sql('shipping_final', engine, if_exists='replace', index=False)

print(f"SUKSES! Total order bersih yang dimasukkan: {len(df_orders_clean)}")
```
